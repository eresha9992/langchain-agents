```python
from langchain.tools import tool,ToolRuntime
@tool
def get_favourite_colour(runtime: ToolRuntime) -> str:
    """Get the user's favourite colour."""
    return runtime.context.favourite_colour

@tool
def get_least_favourite_colour(runtime: ToolRuntime) -> str:
    """Get the user's least favourite colour."""
    return runtime.context.least_favourite_colour
```

```python
agent = create_agent(
    model="groq:llama-3.1-8b-instant",
    tools=[
        get_favourite_colour,
        get_least_favourite_colour
    ],
    context_schema=ColourContext
)
```

```python
response = agent.invoke(
    {
        "messages": [
            HumanMessage(
                content="What is my favourite colour?"
            )
        ]
    },
    context=ColourContext(favourite_colour="green")
)
pprint(response)
```

```python
from dataclasses import dataclass
from pprint import pprint

from langchain.agents import create_agent
from langchain.messages import HumanMessage
from langchain.tools import tool, ToolRuntime
from langgraph.checkpoint.memory import InMemorySaver


@dataclass
class UserInfo:
    user_id: str = "1234"


@tool
def get_user_demographic(runtime: ToolRuntime) -> str:
    """Get demographic information for a user."""

    user_id = runtime.context.user_id

    pprint(runtime.state)

    return f"User {user_id} is from Hyderabad."


agent = create_agent(
    model="groq:llama-3.3-70b-versatile",
    tools=[get_user_demographic],
    context_schema=UserInfo,
    checkpointer=InMemorySaver()
)

config = {
    "configurable": {
        "thread_id": "1"
    }
}

response = agent.invoke(
    {
        "messages": [
            HumanMessage(
                content="What is the demographic information?"
            )
        ]
    },
    context=UserInfo(),
    config=config
)

pprint(response)

print(response["messages"][-1].content)
```

```python
from typing import List, Optional
from dataclasses import dataclass
from pprint import pprint

from dotenv import load_dotenv
from pydantic import BaseModel, Field

from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langchain.tools import tool, ToolRuntime
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver

load_dotenv()

# --------------------------------------------------------------------------
# Constants
# --------------------------------------------------------------------------

DEFAULT_MODEL = "llama-3.3-70b-versatile"
DEFAULT_PERSIST_DIR = "./persona_memory"

# --------------------------------------------------------------------------
# Runtime Context
# --------------------------------------------------------------------------

@dataclass
class UserInfo:
    user_id: str = "default_user"


# --------------------------------------------------------------------------
# Tools
# --------------------------------------------------------------------------

@tool
def save_persona_fact(
    facts: List[str],
    runtime: ToolRuntime,
) -> str:
    """
    Save persona facts.
    """
    user_id = runtime.context.user_id

    print("User ID :", user_id)
    print("Facts   :", facts)

    return f"Saved facts to persona memory for user_id={user_id}"


@tool
def search_persona(
    query: str,
    runtime: ToolRuntime,
    k: int = 5,
) -> str:
    """
    Search persona memory.
    """
    user_id = runtime.context.user_id

    print(f"Searching memory for {user_id}")
    print(f"Query : {query}")

    # Replace with your Chroma/vector search logic
    return ""


PERSONA_TOOLS = [
    save_persona_fact,
    search_persona,
]

# --------------------------------------------------------------------------
# System Prompt
# --------------------------------------------------------------------------

PERSONA_SYSTEM_PROMPT = """
You are a helpful assistant that also maintains a long-term persona memory
about the user you're talking to.

Rules:

- Whenever the user reveals a durable fact about themselves
  (role, company, project, skills, preferences, interests,
  location or goals), call save_persona_fact.

- Save facts as short atomic third-person sentences.

Example:
"Works as a Java Developer."
"Is based in Bangalore."

- Do not save temporary conversation.

- Never save passwords, bank details or sensitive information.

- Before saying you don't know something about the user,
  call search_persona.

- Answer naturally.
"""

# --------------------------------------------------------------------------
# Agent Builder
# --------------------------------------------------------------------------

def build_persona_agent(
    model: Optional[str] = None,
    checkpointer=None,
):
    llm = ChatGroq(
        model=model or DEFAULT_MODEL,
        temperature=0,
    )

    agent = create_agent(
        model=llm,
        tools=PERSONA_TOOLS,
        system_prompt=PERSONA_SYSTEM_PROMPT,
        context_schema=UserInfo,
        checkpointer=checkpointer or InMemorySaver(),
    )

    return agent


# --------------------------------------------------------------------------
# Structured Output
# --------------------------------------------------------------------------

class PersonaExtraction(BaseModel):
    facts: List[str] = Field(
        description="Atomic third-person persona facts."
    )


# --------------------------------------------------------------------------
# One-shot Extraction
# --------------------------------------------------------------------------

def extract_persona_from_text(
    conversation_text: str,
    model: Optional[str] = None,
) -> List[str]:

    llm = ChatGroq(
        model=model or DEFAULT_MODEL,
        temperature=0,
    )

    structured_llm = llm.with_structured_output(PersonaExtraction)

    result = structured_llm.invoke(
        "Extract durable persona facts (role, company, "
        "projects, skills, interests, preferences, location, "
        "goals) about the user.\n\n"
        f"Conversation:\n{conversation_text}"
    )

    return result.facts


# --------------------------------------------------------------------------
# Capture Persona
# --------------------------------------------------------------------------

def capture_persona(
    conversation_text: str,
    user_id: str,
    model: Optional[str] = None,
    persist_dir: str = DEFAULT_PERSIST_DIR,
) -> List[str]:

    facts = extract_persona_from_text(
        conversation_text,
        model=model,
    )

    # Example:
    # store = get_store(user_id, persist_dir)
    # store.save_facts(facts)

    return facts


# --------------------------------------------------------------------------
# Demo
# --------------------------------------------------------------------------

if __name__ == "__main__":

    agent = build_persona_agent()

    config = {
        "configurable": {
            "thread_id": "thread-1235"
        }
    }

    response = agent.invoke(
        {
            "messages": [
                HumanMessage(
                    content="I'm a Lead Data Architect at VidyaSankalp, based in Rajahmundry."
                )
            ]
        },
        context=UserInfo(user_id="1234"),
        config=config,
    )

    print("Agent:")
    print(response["messages"][-1].content)

    response = agent.invoke(
        {
            "messages": [
                HumanMessage(
                    content="What do you know about me?"
                )
            ]
        },
        context=UserInfo(user_id="1234"),
        config=config,
    )

    pprint(response)

    print("Agent:")
    print(response["messages"][-1].content)
```
