```python
from dotenv import load_dotenv

from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

# Load environment variables
load_dotenv()

# Initialize the Groq model
model = init_chat_model(
    model="llama-3.3-70b-versatile",
    model_provider="groq"
)

# System Prompt
system_prompt = (
    "You are a science fiction writer. "
    "Create a capital city on the Moon based on the user's request."
)

# Create Agent
scientific_agent = create_agent(
    model=model,
    system_prompt=system_prompt,
)

# User Question
question = "Create a futuristic capital city on the Moon."

# Invoke Agent
response = scientific_agent.invoke(
    {
        "messages": [question
        ]
    }
)

# Print Final Response
print(response["messages"][-1].content)
```

```python
from dotenv import load_dotenv
from langchain.agents import create_agent

load_dotenv()

system_prompt = """
You are a science fiction writer.

When the user asks for the capital of a planet, invent a fictional space capital city.

Always respond using exactly this structure:

Name: <fictional capital city name>

Location: <where the city is located>

Vibe: <2-3 words>

Economy: <main economy>

Main Industries: <industry 1>, <industry 2>, <industry 3>

Do not add any introduction, conclusion, or extra explanation.
"""

scific_agent = create_agent(
    model="groq:llama-3.3-70b-versatile",
    system_prompt=system_prompt,
)

question = {
    "role": "user",
    "content": "What is the capital of Mars?"
}

response = scific_agent.invoke(
    {
        "messages": [question]
    }
)

print(response["messages"][-1].content)
```
