```python
from urllib import response

from langchain.tools import tool
@tool
def square_root(x: float) -> float:
    """
    calculates the square root of a number
    """
    return x**0.5
```

```python
@tool("square_root")
def tool1(x: float) -> float:
    """
    calculates the square root of a number
    """
    return x**0.5
```

```python
@tool("square_root",description="calculate the square root of a number")
def square_root(x: float) -> float:
    return x**0.5
```

```python
square_root.invoke({"x":467})
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="groq:llama-3.3-70b-versatile",
    tools=[square_root],
    system_prompt="""
You are a mathematics wizard.
Use your tool to calculate the square root of a number.
Always use the square_root tool whenever the user asks for a square root.
"""
)
```

```python
from langchain.messages import HumanMessage
question=HumanMessage(content="what is the square root  of 243?")
response=agent.invoke(
    {"messages":[question]},
)
print(response)

print(response['messages'][-1].content)
```

```python
from langchain.messages import HumanMessage
question=HumanMessage(content="what is the square root  of  the respective numbers which is pipe separated 467|667|234")
response=agent.invoke(
    {"messages":[question]},
)
print(response)
print(response['messages'][-1].content)
```
