```python
from urllib import response

from dotenv import load_dotenv
from ipywidgets import FileUpload
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

load_dotenv()
```

```python
from langchain.agents import create_agent
agent=create_agent(model="groq:qwen/qwen3.6-27b",
                   system_prompt="you are a science fiction writter,create a captial city at the users request")
```

```python
from langchain.messages import HumanMessage
question=HumanMessage(content=[{"type":"text",
                               "text":"what is the captial of Moon?"}])
response=agent.invoke(
    {"messages":[question]}
)
print(response['messages'][1].content)
```

```python
from ipywidgets import FileUpload
from IPython.display import display

uploader = FileUpload(
    accept=".jpg",
    multiple=False
)

display(uploader)
```

```python
print(uploader.value)
```

```python
import base64
uploaded_file=uploader.value[0]
content_mv=uploaded_file['content']
img_bytes=bytes(content_mv)
img_b64=base64.b64encode(img_bytes).decode("utf-8")
```

```python
multimodal_question = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Tell me about Capital"
        },
        {
            "type": "image",
            "base64": img_b64,
            "mime_type": "image/jpeg"
        }
    ]
)

response = agent.invoke(
    {"messages": [multimodal_question]}
)

print(response["messages"][-1].content)
```

```python
import sounddevice as sd
from scipy.io.wavfile import write
import time
from tqdm import tqdm

duration = 5
sample_rate = 44100

print("Recording...")

audio = sd.rec(
    int(duration * sample_rate),
    samplerate=sample_rate,
    channels=1,
    dtype="int16"
)

for _ in tqdm(range(duration * 10)):
    time.sleep(0.1)

sd.wait()

print("Done...")

# Save the recording
write("audio.wav", sample_rate, audio)

print("Audio saved successfully.")
```

```python
from groq import Groq
from dotenv import load_dotenv

load_dotenv()

client = Groq()

with open("audio.wav", "rb") as f:
    transcription = client.audio.transcriptions.create(
        file=("audio.wav", f.read()),
        model="whisper-large-v3"
    )

print(transcription.text)
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage

agent = create_agent(
    model="groq:llama-3.3-70b-versatile"
)

question = HumanMessage(
    content=transcription.text
)

response = agent.invoke(
    {"messages": [question]}
)

print(response["messages"][-1].content)
```
