```python
from dotenv import load_dotenv
from langgraph.checkpoint.memory import InMemorySaver

load_dotenv()
```

```python
#without memory
from langchain.agents import create_agent
agent = create_agent(
    model="groq:llama-3.3-70b-versatile",
)
```

```python
from langchain.messages import HumanMessage
question=HumanMessage(content="Hello My name is Sean and my favourite colour is green")

response=agent.invoke({"messages":[question]})
```

```python
from pprint import pprint
pprint(response)
```

```python
question=HumanMessage(content="what is my favourite colour?")
response=agent.invoke({"messages":[question]})
pprint(response)
```

```python
from langgraph.checkpoint.memory import InMemorySaver
agent=create_agent("groq:llama-3.3-70b-versatile",
                   checkpointer=InMemorySaver()
)
```

```python
question=HumanMessage(content="Hello My name is Sean and my favourite colour is green")
config={"configurable":{"thread_id":"1"}}
response=agent.invoke({"messages":[question]},
                      config,)
pprint(response)
```

```python
question=HumanMessage(content="what is my favourite colour?")
response=agent.invoke({"messages":[question]},config,)
print(response)
```

```python
question=HumanMessage(content="what is my favourite colour?")
config={"configurable":{"thread_id":"1"}}
response=agent.invoke({"messages":[question]},config)
print(response)
```
